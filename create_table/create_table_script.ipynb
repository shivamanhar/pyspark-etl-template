```python
%sql
DROP TABLE IF EXISTS silver_fact_sales;
CREATE TABLE IF NOT EXISTS silver_fact_sales (
  sales_id INT,
  transaction_ts STRING,
  date_key  INT,
  product_key INT,
  customer_key  INT,
  store_key INT,
  payment_key  INT,
  quantity INT,
  unit_price DECIMAL(10,2),
  discount_pct DECIMAL(5,2),
  net_sales_amount DECIMAL(10,2),
  tax_amount DECIMAL(10,2)  
)
```

```python
%sql
DROP TABLE IF EXISTS silver_product;
CREATE TABLE IF NOT EXISTS silver_product (
  product_sk INT,
  product_key INT,
  Product_name STRING,
  category STRING,
  brand STRING,
  price DECIMAL(10,2),
  effective_start_date DATE,
  effective_end_date DATE,
  is_current BOOLEAN,
  surrogate_key STRING
)
```

```python
%sql
DROP TABLE IF EXISTS silver_product_new_cast;
CREATE TABLE IF NOT EXISTS silver_product_new_cast (
  product_sk INT,
  product_key INT,
  product_name  STRING,
  category STRING,
  brand  STRING,
  price DECIMAL(10,2),
  effective_start_date STRING,
  effective_end_date STRING,
  is_current STRING
)
```

```python
%sql
DROP TABLE IF EXISTS silver_customer;
CREATE TABLE IF NOT EXISTS silver_customer (
  customer_key INT,
  customer_name STRING,
  gender  STRING,
  city STRING,
  surrogate_key STRING
)
```
